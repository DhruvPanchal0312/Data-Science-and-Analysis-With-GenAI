## 1. Converting Columns to Datetime (`pd.to_datetime`)
- Use `pd.to_datetime()` to convert strings or numbers into datetime objects.
- Once converted, Pandas recognizes them as dates, enabling powerful operations.

```python
import pandas as pd

products = pd.DataFrame({
    "Product_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer", "Speaker", "Webcam", "SSD"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Office", "Audio", "Accessories", "Storage"],
    "Price": [55000, 700, 1500, 12000, 8500, 2500, 1800, 6500],
    "Purchase_Date": [
        "2026-07-01",      # YYYY-MM-DD
        "02/07/2026",      # DD/MM/YYYY
        "July 3, 2026",    # Month DD, YYYY
        "2026.07.04",      # YYYY.MM.DD
        "05-Jul-2026",     # DD-Mon-YYYY
        "2026/07/06",      # YYYY/MM/DD
        "07-07-2026",      # DD-MM-YYYY
        "08 Aug 2026"      # DD Mon YYYY
    ]
})
products
products.dtypes
```

```python
pd.to_datetime(products['Purchase_Date'] , errors="coerce")
```

## 2. Setting DateTime Index & Resampling
- Set a datetime column as the index with `set_index()`.
- Resample time series using `resample('M').mean()` or other rules ('D', 'W', 'Y').
- Useful for aggregating data by time periods.

```python
products['Purchase_Date'] = pd.to_datetime(products['Purchase_Date'], errors='coerce')

products.set_index('Purchase_Date', inplace=True)

products
```

```python

# 1️⃣ Create a dataset of website visits (string dates + visit counts)
website_visits = pd.DataFrame({
    "Date": [
        "2026-07-01",    # YYYY-MM-DD
        "02/07/2026",    # DD/MM/YYYY
        "July 3, 2026",  # Month DD, YYYY
        "2026.07.04",    # YYYY.MM.DD
        "05-Jul-2026",   # DD-Mon-YYYY
        "2026/07/06",    # YYYY/MM/DD
        "07-07-2026",    # DD-MM-YYYY
        "08 Aug 2026",   # DD Mon YYYY
        "2026-08-09",    # YYYY-MM-DD
        "10/08/2026"     # DD/MM/YYYY
    ],
    "Visits": [120, 150, 180, 170, 210, 190, 250, 230, 275, 300]
})

website_visits





```

```python
website_visits["Date"] = pd.to_datetime(
    website_visits["Date"],
    dayfirst=True, errors="coerce"
)
```

```python
# 3️⃣ Set VisitDate as the index
# Write your code here'


website_visits["Date"] = pd.to_datetime(website_visits["Date"], dayfirst=True)

website_visits.set_index("Date", inplace=True)

print(website_visits)

```

```python
# 4️⃣ Resample to get total visitors per month
# Write your code here
# 4️⃣ Resample to get total visitors per month

monthly_visitors = website_visits.resample("ME").sum()

print(monthly_visitors)
```

```python
# 5️⃣ Resample to get average visitors per week
# Write your code here
monthly_visitors = website_visits.resample("W").mean()

print(monthly_visitors)
```
